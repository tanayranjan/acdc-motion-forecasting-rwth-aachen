## Background and Motivation

Automated driving systems need to anticipate how surrounding traffic participants will move in the near future. Motion forecasting is therefore a key component of scene understanding and downstream decision making. In urban scenes, this is particularly challenging because interactions among vehicles, pedestrians, and cyclists can lead to multiple plausible future trajectories.

Prediction quality can vary strongly across scenario types, agent classes, and traffic densities. If predicted trajectories are incorrect, they become less useful or even harmful for automated driving systems. It is therefore important not only to train or run a forecasting model, but also to analyze error characteristics, uncertainty, and generalization behavior.

<img src="https://www.argoverse.org/assets/images/reference_images/av2_motion_forecasting.png" width="25%" />

The [Argoverse 2 Motion Forecasting dataset](https://argoverse.github.io/user-guide/tasks/motion_forecasting.html) provides trajectories for complex urban traffic scenarios.

This topic is intended to investigate forecasting approaches and to evaluate them.
